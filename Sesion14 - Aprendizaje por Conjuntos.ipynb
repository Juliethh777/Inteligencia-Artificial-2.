# APRENDIZAJE POR CONJUNTOS

## 1. Introducción a conjuntos

El aprendizaje por conjuntos es una técnica de Machine Learning que consiste en combinar varios modelos de aprendizaje para obtener una predicción más precisa y confiable que la que podría conseguir un solo modelo.

La idea principal es que varios modelos trabajen juntos para resolver un mismo problema, aprovechando sus diferentes resultados.

## 2. Clasificadores de votación

Los clasificadores de votación combinan las predicciones de varios modelos y seleccionan una respuesta final según los resultados obtenidos.

### Simple Majority Vote (Votación por mayoría simple)

Consiste en elegir la respuesta que recibe la mayor cantidad de votos de los diferentes modelos.

Ejemplo: si cinco modelos deben clasificar un correo y cuatro dicen que es spam, mientras uno dice que no lo es, la respuesta final será spam.

### Weighted Confidence Levels (Niveles de confianza ponderados)

Consiste en darle mayor importancia a los modelos que tienen un nivel de confianza más alto en sus predicciones.

En lugar de contar todos los votos por igual, se consideran los valores de confianza de cada modelo para obtener una respuesta final.

## 3. Mecánica del muestreo

La mecánica del muestreo consiste en seleccionar diferentes conjuntos de datos para entrenar varios modelos. Esto permite que cada modelo aprenda a partir de distintas muestras y pueda aportar resultados diferentes.

### Bagging (Bootstrap Aggregating)

Consiste en entrenar varios modelos utilizando subconjuntos de datos seleccionados aleatoriamente, normalmente con reemplazo.

Cada modelo aprende de una muestra diferente y después se combinan sus predicciones. Esta técnica ayuda a reducir la variabilidad de los resultados.

### Pasting

Es una técnica similar al Bagging, pero los subconjuntos de datos se seleccionan sin reemplazo. Esto significa que una misma instancia no se repite dentro de la muestra seleccionada.

### Entrenamiento paralelo en subconjuntos aleatorios

En Bagging y Pasting, los diferentes modelos pueden entrenarse de manera independiente y en paralelo, cada uno utilizando su propio subconjunto de datos.

Esto permite aprovechar diferentes partes de la información para realizar una predicción conjunta.

### Instancias fuera de bolsa (Out-of-Bag, OOB)

Son los datos que no fueron seleccionados para entrenar un modelo específico durante el muestreo.

Estas instancias pueden utilizarse para evaluar el rendimiento del modelo sin necesidad de separar otro conjunto de datos exclusivamente para la validación.

## 4. Aleatoriedad en las características

Consiste en seleccionar aleatoriamente algunas características o variables del conjunto de datos para entrenar los modelos.

Esto permite que los modelos aprendan desde diferentes perspectivas, evitando que todos dependan exactamente de las mismas variables.

La aleatoriedad en las características es una idea importante en los Bosques Aleatorios (Random Forests), donde se utilizan diferentes subconjuntos de características para construir los árboles de decisión.

## 5. Bosques aleatorios (Random Forests)

Los bosques aleatorios son un método de aprendizaje por conjuntos que combina varios árboles de decisión.

Cada árbol se entrena utilizando una muestra aleatoria de los datos y, durante su construcción, considera subconjuntos aleatorios de características.

Finalmente, se combinan las predicciones de los árboles mediante votación o promedio, dependiendo del problema.

Su objetivo es obtener resultados más estables y reducir el sobreajuste que puede presentar un árbol individual.

## 6. Algoritmos de Boosting

El Boosting es una técnica de aprendizaje por conjuntos que entrena modelos de manera secuencial, buscando que cada nuevo modelo mejore los resultados de los anteriores.

Los modelos se combinan para construir una predicción final más precisa.

A diferencia del Bagging, que permite entrenar modelos de forma independiente, el Boosting utiliza los resultados de los modelos anteriores para orientar el entrenamiento de los siguientes.

## 7. Stacking

El Stacking es una técnica que combina diferentes modelos de Machine Learning mediante un modelo adicional llamado meta-modelo.

Primero, varios modelos realizan sus predicciones. Después, esas predicciones se utilizan como entrada para el meta-modelo, que aprende cómo combinarlas para producir un resultado final.

Esta técnica permite aprovechar las fortalezas de diferentes tipos de modelos.
