```python
# ============================================================
# RESUMEN - INTELIGENCIA ARTIFICIAL AVANZADA
# Google Colab
# ============================================================

print("=" * 70)
print("        RESUMEN DE INTELIGENCIA ARTIFICIAL AVANZADA")
print("=" * 70)

temas = {
    "1. IA AVANZADA": """
La Inteligencia Artificial avanzada busca modelar patrones complejos
y relaciones no lineales presentes en los datos.

Python es una de las principales herramientas para desarrollar modelos
de IA, mientras que la nube (Cloud) permite almacenar datos, ejecutar
modelos y disponer de recursos computacionales.
""",

    "2. PREPARACIÓN DE DATOS": """
Antes de entrenar un modelo es necesario preparar correctamente los datos.

• Fuga de datos (Data Leakage):
  Ocurre cuando información que no debería estar disponible durante
  el entrenamiento termina influyendo en el modelo.

• Imputación:
  Consiste en completar valores faltantes de los datos.

• Escalamiento:
  Ajusta las variables a una escala similar para facilitar el
  entrenamiento de determinados modelos.

• Reducción de dimensionalidad:
  Disminuye la cantidad de variables manteniendo la información
  más importante.

• Clasificación:
  Permite asignar los datos a diferentes categorías o clases.
""",

    "3. MODELOS SUPERVISADOS": """
Los modelos supervisados aprenden utilizando datos que contienen
una variable objetivo o etiqueta conocida.

Algunas técnicas importantes son:

• Truco Kernel:
  Permite trabajar con relaciones no lineales transformando
  implícitamente los datos a un espacio de mayor dimensión.

• Bagging:
  Entrena varios modelos sobre diferentes muestras de los datos
  y combina sus resultados para mejorar la estabilidad.

• Boosting:
  Construye modelos de manera secuencial, donde cada nuevo modelo
  intenta corregir los errores de los anteriores.
""",

    "4. EVALUACIÓN DE MODELOS": """
La evaluación permite determinar qué tan bien funciona un modelo
y si puede generalizar correctamente a datos nuevos.

• Sesgo (Bias):
  Error causado por un modelo demasiado simple que no logra
  representar adecuadamente los patrones de los datos.

• Varianza (Variance):
  Ocurre cuando el modelo se adapta demasiado a los datos de
  entrenamiento y pierde capacidad de generalización.

• GridSearchCV:
  Prueba diferentes combinaciones de hiperparámetros de forma
  sistemática para encontrar una buena configuración.

• RandomizedSearchCV:
  Prueba combinaciones seleccionadas aleatoriamente, permitiendo
  explorar grandes espacios de hiperparámetros de manera más rápida.
"""
}

# Mostrar resumen
for titulo, contenido in temas.items():
    print("\n" + "=" * 70)
    print(titulo)
    print("=" * 70)
    print(contenido)

print("\n" + "=" * 70)
print("                    FIN DEL RESUMEN")
print("=" * 70)
```
